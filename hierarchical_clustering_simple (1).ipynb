```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering, BisectingKMeans
from scipy.cluster.hierarchy import linkage, dendrogram

X, _ = make_blobs(n_samples=30, centers=3, random_state=7)
plt.scatter(X[:, 0], X[:, 1])
plt.title("Data")
plt.show()
```

## Agglomerative

```python
agg = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = agg.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.title("Agglomerative (3 clusters)")
plt.show()
```

### Dendrogram

```python
Z = linkage(X, method='ward')
plt.figure(figsize=(8, 4))
dendrogram(Z)
plt.title("Dendrogram")
plt.xlabel("Point index")
plt.ylabel("Distance")
plt.show()
```

## Divisive (top-down, via Bisecting K-Means)

```python
div = BisectingKMeans(n_clusters=3, random_state=0)
labels_div = div.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels_div, cmap='viridis')
plt.title("Divisive (3 clusters)")
plt.show()
```
